```python
import pandas as pd
import numpy as np
import datetime as dt
```

```python
# lets do data transformation
# Smart precification
# what is smart precification?
# Smart precification is a method of setting prices based on data analysis and market trends.
df_host = pd.read_json('../../input/data_host.json')
df_host
```

```python
# lets normalize the data
df_host = pd.json_normalize(df_host['info_moveis'])
```

```python
df_host.head()
```

```python
# some columns has columns that need to be removed
# method explode -> explodes the list of values in a column into multiple rows
columns = list(df_host.columns) # transform the columns into a list
columns
```

```python
# we need to use the explode from index 3 to -1
df_host = df_host.explode(columns[3:])
df_host
```

```python
# reset the index
# inplace means that the changes will be applied to the original dataframe, drop=True means that the old index will not be added as a column
df_host.reset_index(inplace=True, drop=True) 
df_host.head()
```

```python
df_host.info()
# object means that the column has many different values
# Type is not specified means that the column has only one value
```

```python
df_host['max_hospedes'] = df_host['max_hospedes'].astype(np.int64)
```

```python
df_host.info()# now we can see that the column max_hospedes is of type int64
```

```python
numerical_cols = ['quantidade_banheiros','quantidade_quartos','quantidade_camas']
```

```python
df_host[numerical_cols] = df_host[numerical_cols].astype(np.int64)
df_host.info()# now we can see that the columns quantidade_banheiros, quantidade_quartos and quantidade_camas are of type int64
```

```python
df_host['avaliacao_geral'] = df_host['avaliacao_geral'].astype(np.float64)
df_host.info()# now we can see that the column avaliacao_geral is of type float64
```

```python
# lets do money transformation
money_cols = ['taxa_deposito','taxa_limpeza','preco']
```

```python
df_host['preco'] = df_host['preco'].apply(lambda x: x.replace('$', '').replace(',','').strip()).astype(np.float64)
```

```python
df_host.info()
```

```python
df_host[['taxa_deposito', 'taxa_limpeza']] = df_host[['taxa_deposito', 'taxa_limpeza']].applymap(lambda x: x.replace('$', '').replace(',','').strip()).astype(np.float64)
```

```python
df_host.info()
```

```python
# lets study the data about the other parameters
# simple tokenization
df_host['descricao_local'] = df_host['descricao_local'].str.lower() # transform the text to lower case
df_host['descricao_local']
```

```python
# lets removing characters using regex
df_host['descricao_local'] = df_host['descricao_local'].str.replace('[^a-zA-Z0-9\-\']',' ', regex=True)
# regex [^a-zA-Z0-9\-\'] means that we will remove all characters that are not letters, numbers, hyphens or apostrophes
```

```python
df_host['descricao_local'] = df_host['descricao_local'].str.replace('(?<!\w)-(?!\w)',' ', regex=True)
# regex (?<!\w)-(?!\w) means that we will remove all hyphens that are not surrounded by letters or numbers
```

```python
df_host['descricao_local']
```

```python
df_host['descricao_local'] = df_host['descricao_local'].str.split()
```

```python
df_host.head()
```

```python
# lets remove keys, punctuation and stop words
df_host['comodidades'] = df_host['comodidades'].str.replace('\{|}|\"', '', regex=True)
```

```python
df_host['comodidades']
```

```python
df_host['comodidades'] = df_host['comodidades'].str.split(',')
```

```python
df_host['comodidades'] 
```
