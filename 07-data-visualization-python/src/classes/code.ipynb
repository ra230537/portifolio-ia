```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../../input/imigrantes_canada.csv')
```

```python
df
```

```python
df.info()
```

```python
# lets change the index
df.set_index('País', inplace=True)
```

```python
df
```

```python
# create range to store years
years = list(map(str, range(1980, 2014)))
```

```python
years
```

```python
brazil = df.loc['Brasil', years]
```

```python
brazil
```

```python
brazil_dict = {
    'year' : brazil.index.tolist(),
    'immigrants': brazil.values.tolist()
}
brazil_df = pd.DataFrame(brazil_dict)
brazil_df
```

```python
# set figure size
plt.figure(figsize=(8, 4)) # inch values
plt.plot(brazil_df['year'], brazil_df['immigrants'])

# set the title and labels
plt.title('Immigratns from Brazil to Canada (1980-2013)')
plt.xlabel('Year')
plt.ylabel('Number of Immigrants')

# values in the x-axis and y-axis
plt.xticks([str(i) for i in range(1980, 2015, 5)])
plt.yticks(np.arange(0, 3000, 500))

#anything after this line is ignored by code
plt.show()
```
